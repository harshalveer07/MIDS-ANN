```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 400)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0, x)

def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, alpha * x)

def softmax(x):
    exp_x = np.exp(x)
    return exp_x / np.sum(exp_x)

plt.figure(figsize=(12, 8))

plt.subplot(3, 2, 1)
plt.plot(x, sigmoid(x))
plt.title("Sigmoid Activation Function")
plt.grid(True)

plt.subplot(3, 2, 2)
plt.plot(x, tanh(x), color='orange')
plt.title("Tanh Activation Function")
plt.grid(True)

plt.subplot(3, 2, 3)
plt.plot(x, relu(x), color='green')
plt.title("ReLU Activation Function")
plt.grid(True)

plt.subplot(3, 2, 4)
plt.plot(x, leaky_relu(x), color='red')
plt.title("Leaky ReLU Activation Function")
plt.grid(True)

plt.subplot(3, 2, 5)
plt.plot(x, softmax(x), color='purple')
plt.title("Softmax Activation Function (1D)")
plt.grid(True)

plt.suptitle("Activation Functions in Neural Networks\nHarshal Veer - Roll No: 75", fontsize=14)
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```
